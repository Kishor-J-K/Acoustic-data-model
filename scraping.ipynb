```python
def function(links, paths):
    base_url = links
    
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(base_url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    audio_link = None

    # Look for audio file links
    for link in soup.find_all('a', href=True):
        if link['href'].endswith(('.aiff', '.wav', '.mp3')):
            audio_link = link['href']
            if audio_link:
                if not audio_link.startswith('http'):
                    audio_link = "https://whoicf2.whoi.edu" + audio_link

                audio_data = requests.get(audio_link, headers=headers).content
                filename = audio_link.split('/')[-1]
                filepath = os.path.join(paths, filename)

                with open(filepath, 'wb') as f:
                    f.write(audio_data)

                print(f"Downloaded: {filename}")
        
```
